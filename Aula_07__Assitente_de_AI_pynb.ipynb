```python
# Solicita ao usuário que digite o primeiro número e o converte para float
num1 = float(input("Digite o primeiro número: "))
# Solicita ao usuário que digite o segundo número e o converte para float
num2 = float(input("Digite o segundo número: "))

# Exibe as opções de operação para o usuário
print("Escolha a operação:")
print("1 - Somar")
print("2 - Subtrair")
print("3 - Multiplicar")
print("4 - Dividir")

# Solicita ao usuário que escolha uma opção de operação
opcao = input("Digite a opção (1/2/3/4): ")

# Verifica a opção escolhida pelo usuário e executa a operação correspondente
if opcao == '1':
    # Executa a soma dos dois números
    resultado = num1 + num2
    operacao_nome = "Soma"
elif opcao == '2':
    # Executa a subtração dos dois números
    resultado = num1 - num2
    operacao_nome = "Subtração"
elif opcao == '3':
    # Executa a multiplicação dos dois números
    resultado = num1 * num2
    operacao_nome = "Multiplicação"
elif opcao == '4':
    # Verifica se o segundo número não é zero antes de realizar a divisão
    if num2 != 0:
        # Executa a divisão dos dois números
        resultado = num1 / num2
        operacao_nome = "Divisão"
    else:
        # Exibe uma mensagem de erro se a divisão por zero for tentada
        print("Erro: Divisão por zero não é permitida.")
        resultado = "Indefinido"
        operacao_nome = "Divisão"
else:
    # Se a opção for inválida, exibe uma mensagem de erro
    print("Opção inválida.")
    resultado = None
    operacao_nome = "Nenhuma"

# Se um resultado válido foi calculado, exibe-o ao usuário
if resultado is not None:
    print(f"Resultado da {operacao_nome}: {resultado}")
```

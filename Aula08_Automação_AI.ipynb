## Organização Automática de Arquivos por Extensão

Este programa Python irá organizar arquivos em um diretório especificado, movendo-os para subpastas criadas com base em suas extensões (ex: `.pdf`, `.jpg`). Utilizaremos as bibliotecas `os` para interagir com o sistema operacional (listar arquivos, criar diretórios) e `shutil` para mover os arquivos.

### Como usar:

1.  **Defina o `target_directory`**: Altere a variável `target_directory` no próximo bloco de código para o caminho da pasta que você deseja organizar. Certifique-se de que o caminho está correto.
2.  **Execute o código**: Execute o bloco de código para iniciar a organização.

**Atenção**: É altamente recomendável **fazer um backup** de seus arquivos ou **testar este script em uma pasta de exemplo** antes de executá-lo em diretórios importantes, pois ele move arquivos permanentemente.

```python
import os
import shutil

def organize_files(directory_path):
    """
    Organiza arquivos em um diretório especificado, movendo-os para subpastas
    baseadas em suas extensões.

    Args:
        directory_path (str): O caminho para o diretório a ser organizado.
    """
    # Verifica se o diretório existe
    if not os.path.isdir(directory_path):
        print(f"Erro: O diretório '{directory_path}' não existe.")
        return

    print(f"Organizando arquivos em: {directory_path}")

    # Itera sobre todos os itens no diretório especificado
    for filename in os.listdir(directory_path):
        file_path = os.path.join(directory_path, filename)

        # Verifica se o item é um arquivo (e não um diretório ou link simbólico)
        if os.path.isfile(file_path):
            # Obtém o nome base do arquivo e a extensão. Ex: 'documento.pdf' -> ('documento', '.pdf')
            _, extension = os.path.splitext(filename)

            # Se o arquivo não tiver extensão (ex: 'LICENSE'), atribui uma categoria padrão em português
            if not extension:
                folder_name = "ARQUIVOS_SEM_EXTENSAO"
            else:
                # Remove o ponto da extensão e a converte para maiúsculas para o nome da pasta em português
                # Ex: '.pdf' -> 'PDF_ARQUIVOS'
                folder_name = f"{extension[1:].upper()}_ARQUIVOS"

            # Constrói o caminho completo para a pasta de destino
            destination_folder = os.path.join(directory_path, folder_name)

            # Cria a pasta de destino se ela ainda não existir
            if not os.path.exists(destination_folder):
                os.makedirs(destination_folder)
                print(f"  Pasta criada: {destination_folder}")

            # Move o arquivo para a pasta de destino
            try:
                shutil.move(file_path, os.path.join(destination_folder, filename))
                print(f"  Movido: '{filename}' para '{folder_name}'")
            except Exception as e:
                print(f"  Erro ao mover '{filename}': {e}")
    print("Organização de arquivos concluída!")
```

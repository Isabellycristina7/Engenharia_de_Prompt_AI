```python
# Solicita a temperatura do sistema ao usuário e converte para um número inteiro
temperatura = int(input("Informe a temperatura do sistema: "))

# Verifica se a temperatura é maior que 30 graus Celsius
if temperatura > 30:
    # Se for maior que 30, exibe um alerta de superaquecimento
    print("ALERTA: Superaquecimento! Temperatura atual: " + str(temperatura) + "°C")
else:
    # Caso contrário, informa que a temperatura está normal
    print("Temperatura normal: " + str(temperatura) + "°C")
```
